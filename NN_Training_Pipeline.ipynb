```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
# Now we have all the data in required format
# Now we will define our model
# we have 30 features [cols] so here we will take 30 weights an 1 bias [here we have simple neuron or 1 neuron]
```

```python
class NNet():
  def __init__(self, X):

    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
    # here X.shape[1] will give 30 so it will be a 30*1 metrices
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)
    # since we have one neuron so only 1 bias is required

  # define forward pass
  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    ypred = torch.sigmoid(z)
    return ypred

  # define loss function [since this is binary classification so we will use binary cross entropy]
  def loss_fun(self, ypred, y):
    # Clamp predictions to avoid the log(0)
    epsilon = 1e-7
    ypred = torch.clamp(ypred, epsilon, 1-epsilon)

    # calculate loss
    loss = -(y_train_tensor*torch.log(ypred) + (1 + y_train_tensor) * torch.log(1 - ypred)).mean()
    return loss
```

```python
learning_rate = 0.1
epochs = 25
```

```python
# define the model
model = NNet(X_train_tensor)
model.weights

```

```python
model.bias
```

```python
# model training

for epoch in range(epochs):
  # forward pass
  ypred = model.forward(X_train_tensor)
  # print(ypred)

  # calculate loss
  loss =  model.loss_fun(ypred, y_train_tensor)
  # print(f'Epoch : {epoch + 1}, Loss : {loss.item()}')
  # backward pass
  # parameters update

  # backword pass
  loss.backward()

  # weights update
  # here we don't need gradient tracking
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # make gradients 0 to avoid the accumulation
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  print(f'Epoch : {epoch + 1}, Loss : {loss.item()}')
```
