Full-Batch Gradient Descent — Summary

Suppose our network has:

18 weights
4 biases
22 total parameters
100 training samples
What happens in one epoch?
Entire X_train
      ↓
Forward Pass
      ↓
100 predictions
      ↓
Calculate loss for each sample
      ↓
Average the 100 losses
      ↓
One overall loss
      ↓
loss.backward()
      ↓
22 individual gradients
      ↓
Update all 22 parameters
Important distinction

The same overall loss is used as the objective for every parameter.

But each parameter gets its own gradient:

∂w

∂L
	​

2
	​

∂L
	​

∂b
4
	​

∂L
	​


Therefore, each parameter gets a different update.

                 SAME OVERALL LOSS
                        ↓
          ┌─────────────┼─────────────┐
          ↓             ↓             ↓
         w₁            w₂            b₁
          ↓             ↓             ↓
      gradient       gradient       gradient
          ↓             ↓             ↓
      update w₁     update w₂     update b₁
Why average the loss?

The average loss gives us one measure of how well the entire model is performing across the training dataset.

The gradients then tell us:

How does changing each individual parameter affect that overall loss?

So:

Same objective→different gradient per parameter→different update per parameter
	​

What is your current approach called?

Full-Batch Gradient Descent

Entire training dataset → one forward/backward pass
One overall loss
All parameters updated once
Therefore, 1 epoch = 1 parameter-update step

X_test and y_test are not used for training; they're used afterward to evaluate the trained model.

```python
from sklearn.datasets import make_classification
import torch
```

```python
# Step 1: Create a synthetic classification dataset using sklearn
X, y = make_classification(
    n_samples=10,       # Number of samples
    n_features=2,       # Number of features
    n_informative=2,    # Number of informative features
    n_redundant=0,      # Number of redundant features
    n_classes=2,        # Number of classes
    random_state=42     # For reproducibility
)
```

```python
# Convert this data to pytorch tensors
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):

  def __init__(self, features, labels):
    self.features = features
    self.labels = labels

  def __len__(self):
    return self.features.shape[0]

  def __getitem__(self, index):
    return self.features[index], self.labels[index]
```

```python
dataset = CustomDataset(X, y)
```

```python
dataset.features
```

```python
dataloader = DataLoader(dataset, batch_size=2, shuffle=False)
```

```python
for batch_features, batch_labels in dataloader:

  print(batch_features)
  print(batch_labels)
  print("-"*50)
```
